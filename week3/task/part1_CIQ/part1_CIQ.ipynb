```python
def calculate_l2_norm(original, quantized):
    norm = np.sqrt(np.sum((original - quantized) ** 2))
    return norm
```

```python
# Load images
lena_path = './images_CIQ/lena.png'

lena = cv2.imread(lena_path).astype(np.float32) / 255.0

# Parameters
k = 16
max_iterations = 100
visualize = True
```
